```python
import re #regular expression 
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer


stop_words=set(stopwords.words('english'))
ps=PorterStemmer()
lemmatizer=WordNetLemmatizer()
```

```python
text_data = [
    "I LOVE Python!!!",
    "Machine Learning is AMAZING.",
    "Call me at 9876543210.",
    "Visit https://google.com now.",
    "My email is abc@gmail.com.",
    "<h1>Hello World</h1>",
    "Python version 3.13 released.",
    "Data Science is fun 😊",
    "Natural Language Processing (NLP) is awesome!!",
    "Artificial Intelligence is changing the world.",
    "Running, playing, studying and reading books.",
    "Children are eating chocolates.",
    "The cats were running in the garden.",
    "Don't forget to complete your assignments.",
    "Today's weather is very pleasant."
]
```

```python
r=[]
for sentence in text_data:
    original=sentence
    #print(original)
    #1 lowercase
    sentence=sentence.lower()

    #2 remove html tags
    sentence=re.sub(r'<.*?>',' ',sentence)

    #3 remove url
    sentence = re.sub(r'https?://\S+', '', sentence)

    #4 removal email
    sentence = re.sub(r'\S+@\S+', '', sentence)
    #5 removo numbers
    sentence=re.sub(r'\d+','',sentence)

    #6 removo punctuvation
    sentence=sentence.translate(str.maketrans('','',string.punctuation)) # replace,replace with ,delete

    # removo extra space
    sentence = re.sub(r'\s+', ' ', sentence).strip()

    #8 tokenization
    words=word_tokenize(sentence)

    #9 removo stopwords
    words=[word for word in words if word not in stop_words]

    #10 stemmimg
    stem_words=[ps.stem(word) for word in words]

    #11 lemmatization 
    lemma_words=[lemmatizer.lemmatize(word) for word in words]

    r.append({
        'orignal':original,
        'tokens':words,
        'stemmed': stem_words,
        'Lemmatized':lemma_words,
        
    })

df=pd.DataFrame(r)

df




    









    

```
