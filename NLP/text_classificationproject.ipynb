```python
import pandas as pd 


df=pd.read_csv(r"C:\Users\vilas\Downloads\archive (1)\spam (or) ham.csv")
df

df['Class'].unique()
df['Class'].value_counts()
df.shape
df.isnull().sum()
df.duplicated().sum()

df
```

```python
# data understanding
df.columns.tolist()
df['Class'].unique().tolist()
df['Class'].value_counts()
```

```python
#missing value removo

df=df.dropna()
df.shape
```

```python
# data preprossesing 
#1- reovo punctuvation 
#2- removo stopword
#3-steammmg
import re
import nltk 
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps=PorterStemmer()

corpus=[]
for i in range(len(df)):
    r=re.sub('[^a-zA-Z]'," ",df['sms'][i])
    r=r.lower()
    r=r.split()
    r=[ps.stem(word) for word in r if not word in stopwords.words('english')] # reduce no of columns  then use stemmimg  and stopword 
    r=" ".join(r)
    corpus.append(r)





```

```python
df.shape
```

```python
# vectorization  is on structure data
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer(max_features=None) #word(feature) not sentence  
X=cv.fit_transform(corpus).toarray()
X.shape

```

```python
# label encoding is unstructure data
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Class']=le.fit_transform(df[['Class']])
y=df['Class']
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.8,random_state=True)


```

```python
# multclass naive bayes 
#train the model using naive_bayes 
from sklearn.naive_bayes import MultinomialNB
model=MultinomialNB()
model.fit(X_train,y_train)
```

```python
# evalute your machine
ypred_train=model.predict(X_train)
ypred_test=model.predict(X_test)

from sklearn.metrics import accuracy_score
acc_train=accuracy_score(y_train,ypred_train)
acc_test=accuracy_score(y_test,ypred_test)

print('acc_train:',acc_train)
print('acc_test:',acc_test)


```

```python
#cross validation score
from sklearn.model_selection import cross_val_score

cross_val_score(model,X,y,cv=5).mean()
```
