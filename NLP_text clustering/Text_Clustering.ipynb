```python
import pandas as pd
df=pd.read_csv(r"C:\Users\vilas\Downloads\Telegram Desktop\npr.csv")

df
df.isnull().sum()
df.value_counts()
```

```python
#preprocessing 
import nltk 
import re
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords  
wnl=WordNetLemmatizer()

corpus=[]
for i in range(len(df)):
    r=re.sub('[^a-zA-Z]'," ",df['Article'][i])
    r=r.lower()
    r=r.split()
    r=[wnl.lemmatize(word) for word in r if not word in set(stopwords.words('english'))]
    r=" ".join(r)
    corpus.append(r)

print(corpus)
   


```

```python
#tf_idffrom sklearn.feature_extraction.text import  TfidfVectorizer
tf=TfidfVectorizer()
bow=pd.DataFrame(tf.fit_transform(corpus).toarray(),columns=tf.get_feature_names_out())

bow
```

```python
# vectorization 
from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer()
X=cv.fit_transform(corpus).toarray()
X
#unique word create 12008

# X input

```

```python
from sklearn.decomposition import LatentDirichletAllocation
model=LatentDirichletAllocation(n_components=3)
model.fit(X)  # calculating distance from word to word 


```

```python
#Finds the probability that each document belongs to each topic.
topic_result=model.transform(X) # showing the topic probability 

topic_result[0]  #Show the topic probabilities for the first document.
topic_result[0].argmax()
#array([5.43149475e-04, 5.40813238e-04, 9.98916037e-01])
# output is index mens 2 mens =9.98916037e-01 
```

```python
df['group']=topic_result.argmax(axis=1)
df
```
